- underfit(欠拟合)/high bias(高偏差)

    对训练集的拟合程度较差，通常可能由训练次数过少或特征数太少导致
    ![alt text](note.assets/image.png)
- overfit(过拟合)/high variance(高方差)

    过于贴合训练集，泛化能力弱，通常可能由训练次数过多或特征数太多导致
    ![alt text](note.assets/image-1.png)
    这里不同的次幂项也代表不同的特征
- 应该要达到刚刚好
![alt text](note.assets/image-2.png)
分类问题中也有这三种情况

![alt text](note.assets/image-3.png)
## addressing overfit
1. 获取更多数据进行训练
2. 做合适的特征选择
3. 正则化，即减小一些特征的影响但又不彻底消除它们，只是减小这些特征前面的系数w

## Regularization(正则化)
![alt text](note.assets/image-4.png)
通过在loss函数中引入$\lambda$项，训练过程中会考虑减小w的值，因为我们不知道哪些特征更重要，所以我们一视同仁的对这些w进行控制，公式中给第二项也加上m分母可以在数据集数量变化时(m变化)也能较好的满足loss的最小化，$\lambda$的取值决定了你对这两项的权衡，考虑极端情况，若$\lambda$过小，则你仍然只考虑更好地拟合训练集，导致overfit,若过大，则w都会变得特别小，只在考虑最小化第二项，导致第一项仍然较大，也就是underfit.注意这里是否加$\frac{\lambda}{2m}b^2$项对拟合效果的影响很小，所以可加可不加
