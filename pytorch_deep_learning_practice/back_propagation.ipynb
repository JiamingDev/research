```python
#batch_size=1
import torch
import matplotlib.pyplot as plt
x_data=[1.,2.,3.]
y_data=[2.,4.,6.]
w=torch.tensor(1.,requires_grad=True)
def loss(x,y):
    y_pred=x*w
    return (y_pred-y)**2
L=[]
for _ in range(100):
    sum=0
    for x,y in zip(x_data,y_data):
        l=loss(x,y)
        sum+=l.item()
        l.backward()
        w.data-=0.01*w.grad.data
        w.grad.data.zero_()
    L.append(sum/3)
print(w.item())
plt.plot(L)
plt.xlabel('epoch')
plt.ylabel('loss')
plt.show()

    
```

```python
#换为y_hat=w1*x^2+w2*x+b
import torch
import matplotlib.pyplot as plt
w1=torch.tensor(1.,requires_grad=True)
w2=torch.tensor(1.,requires_grad=True)
b=torch.tensor(1.,requires_grad=True)
def loss2(x,y):
    y_pred=w1*x**2+w2*x+b
    return(y_pred-y)**2
L=[]
for _ in range(100):
    sum=0
    for x,y in zip(x_data,y_data):
        l=loss2(x,y)
        sum+=l.item()
        l.backward()
        with torch.no_grad():
            w1-=0.001*w1.grad
            w2-=0.001*w2.grad
            b-=0.001*b.grad
    
        for t in[w1,w2,b]:
            t.grad.zero_()
    L.append(sum/3)
plt.plot(L)
plt.xlabel('epoch')
plt.ylabel('loss')
plt.show()


```
