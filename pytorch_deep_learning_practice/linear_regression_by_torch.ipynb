```python
import torch
x_data=torch.tensor([[1.],[2.],[3.]])#3*1矩阵
y_data=torch.tensor([[2.],[4.],[6.]])#3*1矩阵
class ModelLinear(torch.nn.Module):
    #构造函数
    def __init__(self):
        #父类先初始化
        super().__init__()
        #创建实例对象，输入1个特征，输出1个特征，
        # y=w*x+b,自动帮你创建w和b,
        # 第三个参数默认是bias=True即默认有b
        self.linear=torch.nn.Linear(1,1)
    #前馈函数，本质是覆盖父类中已有的forward，所以名字不能变
    # 而且父类中的__call__已经使得创建的对象callable，
    # 且call时会自动调用forward函数
    def forward(self,x):
        return self.linear(x)
model=ModelLinear()#实例化
#计算loss函数，默认取mean
criterion=torch.nn.MSELoss()
#优化器：根据梯度更新参数
#第一个参数告诉它要更新哪些参数
#第二个是学习率
optimizer=torch.optim.SGD(model.parameters(),lr=0.01)
for epoch in range(1000):
    #前馈
    y_pred=model(x_data)
    loss=criterion(y_pred,y_data)
    print(epoch,loss.item())
    #清零
    optimizer.zero_grad()
    #反向传播
    loss.backward()
    #更新
    optimizer.step()
print('w=',model.linear.weight.item())
print('b=',model.linear.bias.item())
x_test=torch.tensor([[4.]])
y_test=model(x_test)
print(y_test.data)
```
